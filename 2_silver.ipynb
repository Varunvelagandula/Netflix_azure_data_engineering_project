# Lookup variables

### Parameters

```python
dbutils.widgets.text("sourcefolder", "Data")
dbutils.widgets.text("targetfolder", "Data")
```

### variables

```python
var_src_folder = dbutils.widgets.get("sourcefolder")
var_target_folder = dbutils.widgets.get("targetfolder")
```

### Dynamic read and write

```python
df = spark.read.format("csv")\
    .option("header", "true")\
        .option("inferSchema", "true")\
            .load(f"abfss://bronze@dlnetflix.dfs.core.windows.net/{var_src_folder}")

df.write.format("delta")\
    .mode("overwrite")\
        .option("path",f"abfss://silver@dlnetflix.dfs.core.windows.net/{var_target_folder}")\
            .save()
```
